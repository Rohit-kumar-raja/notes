## what is autogard ?

Autograd = Automatic Gradient Computation Engine

It automatically computes gradients (derivatives) of tensors, which are required for training neural networks via backpropagation.

#### How it works (conceptually)

Autograd builds a dynamic computation graph during the forward pass:

You perform tensor operations

PyTorch records operations in a graph

When you call .backward(), it:

Traverses the graph backward

Computes gradients using chain rule

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2 + 3 * x + 1   # computation tracked

y.backward()             # compute gradient

print(x.grad)            # dy/dx = 2x + 3 → 7
```
