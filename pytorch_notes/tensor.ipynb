# Creating Tensor and practical from here ""

```python
import torch
```

```python
a = torch.tensor([1, 2, 3])
print(a)
```

```python
a = torch.empty(2, 2, 2)
print(a)
```

```python
torch.manual_seed(100)
torch.rand(2, 3)
```

```python
import random

random.seed(100)
random.random() * 10  # core part of seed come from here
```

```python
torch.range(1, 10, 2)
```

```python
torch.linspace(1, 10, 9)
```

```python
torch.eye(50)
```

```python
torch.tensor(
    [1, 2, 3, 4, True, False]
)  # tensor([1, 2, 3, 4, 1, 0], dtype=torch.uint8)  # True is converted to 1 and False is converted to 0
```

```python
torch.full_like(torch.tensor([1, 2, 3]), 5).shape  # tensor([5, 5, 5])
```

```python
x = torch.tensor([1, 2, 3], dtype=torch.float32)
```

```python
x.to(torch.int32)
```

```python
print("Tensor Data Types:")
print("{:<20} {:<20}".format("Data Type", "Description"))
print("{:<20} {:<20}".format("torch.float32", "32-bit floating point"))
print("{:<20} {:<20}".format("torch.float64", "64-bit floating point"))
print("{:<20} {:<20}".format("torch.int32", "32-bit integer"))
print("{:<20} {:<20}".format("torch.int64", "64 bit integer"))
print("{:<20} {:<20}".format("torch.uint8", "8-bit unsigned integer"))
print("{:<20} {:<20}".format("torch.bool", "Boolean (True/False)"))
```

### Mathematic Calculations 

```python
a = torch.tensor([[1, 2, 3, 4000], [1, 2, 3, 4], [1, 2, 3, 4]])
b = torch.tensor([[2, 4, 5, 6000], [2, 4, 5, 6], [2, 4, 5, 6]])
```

```python
# add
a + 2
```

```python
# sub
a - 2
```

```python
# division
a / 2
```

```python
# power
a**2
```

```python
# mod
a % 2
```

 ####  Mathematic operations between more than one tensor

```python
a-b
```

```python
a/b
```

```python
a*b
```

```python
a**b   # first row of third element will the zero because of data overflow range 
```
